# پیش‌بینی شانس حمله قلبی — مقایسه چهار مدل Classification
در این پروژه با استفاده از داده‌های پزشکی بیماران، مدل‌های **Logistic Regression**، **KNN**، **Decision Tree** و **SVM** را می‌سازیم، عملکردشان را با هم مقایسه می‌کنیم و بهترین مدل را انتخاب می‌کنیم.
دیتاست: `heart.csv` — ۳۰۳ رکورد بیمار، ستون هدف `output` (۰ = شانس کم، ۱ = شانس زیاد).

## ۰. ایمپورت کتابخانه‌ها

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.metrics import (accuracy_score, precision_score, recall_score,
                              f1_score, confusion_matrix, classification_report)
import joblib
import warnings
warnings.filterwarnings('ignore')

%matplotlib inline
sns.set_style('whitegrid')
RANDOM_STATE = 42
```

## ۱. بارگذاری داده و نگاه اول
خواندن فایل CSV و بررسی ساختار کلی، تایپ ستون‌ها و چند ردیف نمونه.

```python
df = pd.read_csv('heart.csv')
print("Shape:", df.shape)
df.head()
```

```python
df.info()
```

## ۲. تحلیل اکتشافی داده (EDA)
بررسی مقادیر گمشده، رکوردهای تکراری و توزیع کلی متغیرها.

```python
print("مقادیر گمشده در هر ستون:")
print(df.isnull().sum())
print("\nتعداد رکوردهای کاملاً تکراری:", df.duplicated().sum())
df.describe().T
```

**نکته مهم:** طبق توضیح دیتاست، ستون `caa` باید بین ۰ تا ۳ و ستون `thall` باید بین ۱ تا ۳ باشد. بیایید مقادیر خارج از این بازه را پیدا کنیم — این نوعی «داده نامعتبر پنهان» است که با `isnull()` دیده نمی‌شود.

```python
print("مقادیر caa:\n", df['caa'].value_counts().sort_index())
print("\nمقادیر thall:\n", df['thall'].value_counts().sort_index())
invalid_count = ((df['caa'] == 4) | (df['thall'] == 0)).sum()
print(f"\nتعداد رکوردهای دارای مقدار نامعتبر: {invalid_count}")
```

### توزیع متغیر هدف (target)
مهم است بدانیم دیتاست چقدر متوازن (balanced) است، چون این موضوع روی انتخاب معیار ارزیابی و روش Split اثر می‌گذارد.

```python
ax = sns.countplot(data=df, x='output', hue='output', palette='Set2', legend=False)
ax.set_xticks([0, 1])
ax.set_xticklabels(['شانس کم (0)', 'شانس زیاد (1)'])
plt.title('توزیع کلاس هدف')
plt.xlabel('')
plt.ylabel('تعداد بیماران')
plt.show()
print(df['output'].value_counts(normalize=True).round(3))
```

### همبستگی بین ویژگی‌ها
نقشه حرارتی همبستگی به ما کمک می‌کند بفهمیم کدام ویژگی‌ها رابطه قوی‌تری با `output` دارند.

```python
plt.figure(figsize=(11, 9))
sns.heatmap(df.corr(numeric_only=True), annot=True, fmt='.2f', cmap='coolwarm', center=0)
plt.title('ماتریس همبستگی ویژگی‌ها')
plt.show()
```

## ۳. پاکسازی داده (Data Cleaning)
حذف رکورد تکراری و رکوردهایی که مقدار نامعتبر در `caa` یا `thall` دارند.

```python
before = df.shape[0]
df = df.drop_duplicates()
df = df[(df['caa'] != 4) & (df['thall'] != 0)].reset_index(drop=True)
after = df.shape[0]
print(f"تعداد رکورد قبل از پاکسازی: {before}")
print(f"تعداد رکورد بعد از پاکسازی: {after}  (حذف شد: {before - after})")
```

## ۴. آماده‌سازی ویژگی‌ها (Preprocessing)
چند ستون مثل `cp`، `restecg`، `slp` و `thall` اعداد صحیح‌اند اما در واقع دسته‌بندی (Nominal Categorical) هستند، نه عدد قابل مقایسه — یعنی cp=3 «بزرگ‌تر» از cp=1 نیست، فقط دسته متفاوتی است.
برای مدل‌هایی مثل Logistic Regression، KNN و SVM که بر پایه فاصله/وزن خطی کار می‌کنند، این ستون‌ها را با One-Hot Encoding به چند ستون صفر و یک تبدیل می‌کنیم تا مدل رابطه ترتیبی غلط بین دسته‌ها استنباط نکند.

```python
nominal_cols = ['cp', 'restecg', 'slp', 'thall']
df_encoded = pd.get_dummies(df, columns=nominal_cols, drop_first=True)

X = df_encoded.drop('output', axis=1)
y = df_encoded['output']
print("تعداد ویژگی‌ها بعد از encoding:", X.shape[1])
X.head()
```

### تقسیم داده به Train و Test
از `stratify=y` استفاده می‌کنیم تا نسبت دو کلاس در هر دو مجموعه train و test مثل داده اصلی حفظ شود.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
)
print("Train:", X_train.shape, " Test:", X_test.shape)
```

### مقیاس‌بندی ویژگی‌ها (Feature Scaling)
KNN، SVM و Logistic Regression به مقیاس ویژگی‌ها حساس‌اند (مثلاً `chol` بین ۱۰۰ تا ۵۰۰ ولی `oldpeak` بین ۰ تا ۶ است)، پس با `StandardScaler` همه را هم‌مقیاس می‌کنیم.
نکته کلیدی: `scaler` فقط روی `X_train` فیت (fit) می‌شود تا اطلاعاتی از داده تست نشتی (Data Leakage) به مدل نرود.

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## ۵. ساخت و ارزیابی مدل‌ها
یک تابع کمکی می‌سازیم که هر مدل را آموزش می‌دهد، پیش‌بینی می‌گیرد، معیارهای Accuracy، Precision، Recall و F1-Score را حساب می‌کند و نتیجه را در جدول مقایسه ذخیره می‌کند.
در مسئله پزشکی مثل این، **Recall کلاس ۱ (شانس زیاد)** اهمیت زیادی دارد: هزینه از‌دست‌دادن یک بیمار پرخطر (False Negative) بیشتر از هشدار اشتباه به یک بیمار سالم است.

```python
results = []

def evaluate_model(name, model, X_tr, X_te, y_tr, y_te):
    model.fit(X_tr, y_tr)
    y_pred = model.predict(X_te)
    metrics = {
        'Model': name,
        'Accuracy': accuracy_score(y_te, y_pred),
        'Precision': precision_score(y_te, y_pred),
        'Recall': recall_score(y_te, y_pred),
        'F1-Score': f1_score(y_te, y_pred),
    }
    results.append(metrics)
    print(f"--- {name} ---")
    for k, v in metrics.items():
        if k != 'Model':
            print(f"{k}: {v:.3f}")
    return model, y_pred
```

### مدل ۱: Logistic Regression
مدل خطی و ساده، نقطه شروع خوب برای مقایسه با مدل‌های پیچیده‌تر.

```python
log_reg = LogisticRegression(max_iter=1000, random_state=RANDOM_STATE)
log_reg, y_pred_lr = evaluate_model('Logistic Regression', log_reg,
                                     X_train_scaled, X_test_scaled, y_train, y_test)
```

### مدل ۲: K-Nearest Neighbors (KNN)
چون عملکرد KNN شدیداً به مقدار `k` وابسته است، با `GridSearchCV` و اعتبارسنجی متقابل (Cross-Validation) بهترین مقدار k را بین چند گزینه فرد پیدا می‌کنیم.

```python
knn_params = {'n_neighbors': list(range(3, 22, 2))}
knn_grid = GridSearchCV(KNeighborsClassifier(), knn_params, cv=5, scoring='f1')
knn_grid.fit(X_train_scaled, y_train)
print("بهترین k:", knn_grid.best_params_['n_neighbors'])

best_knn, y_pred_knn = evaluate_model('KNN', knn_grid.best_estimator_,
                                       X_train_scaled, X_test_scaled, y_train, y_test)
```

### مدل ۳: Decision Tree
درخت تصمیم نیازی به مقیاس‌بندی ندارد (چون بر پایه آستانه‌گذاری روی هر ویژگی کار می‌کند، نه فاصله). با `GridSearchCV` عمق درخت و حداقل نمونه هر برگ را تنظیم می‌کنیم تا از Overfitting جلوگیری شود.

```python
dt_params = {'max_depth': [2, 3, 4, 5, 6, 7, 8, None], 'min_samples_leaf': [1, 2, 4]}
dt_grid = GridSearchCV(DecisionTreeClassifier(random_state=RANDOM_STATE), dt_params, cv=5, scoring='f1')
dt_grid.fit(X_train, y_train)
print("بهترین پارامترها:", dt_grid.best_params_)

best_dt, y_pred_dt = evaluate_model('Decision Tree', dt_grid.best_estimator_,
                                     X_train, X_test, y_train, y_test)
```

### مدل ۴: Support Vector Machine (SVM)
چند مقدار برای `C` (میزان تحمل خطا) و دو نوع kernel (`linear` و `rbf`) را با `GridSearchCV` امتحان می‌کنیم تا بهترین ترکیب پیدا شود.

```python
svm_params = {'C': [0.1, 1, 10], 'kernel': ['linear', 'rbf'], 'gamma': ['scale']}
svm_grid = GridSearchCV(SVC(probability=True, random_state=RANDOM_STATE), svm_params, cv=5, scoring='f1')
svm_grid.fit(X_train_scaled, y_train)
print("بهترین پارامترها:", svm_grid.best_params_)

best_svm, y_pred_svm = evaluate_model('SVM', svm_grid.best_estimator_,
                                       X_train_scaled, X_test_scaled, y_train, y_test)
```

## ۶. مقایسه نهایی مدل‌ها
همه معیارها را کنار هم در یک جدول و یک نمودار میله‌ای می‌بینیم.

```python
results_df = pd.DataFrame(results).sort_values('F1-Score', ascending=False).reset_index(drop=True)
results_df
```

```python
results_df.set_index('Model')[['Accuracy', 'Precision', 'Recall', 'F1-Score']].plot(
    kind='bar', figsize=(10, 6), rot=0, colormap='viridis'
)
plt.title('مقایسه معیارهای ارزیابی بین مدل‌ها')
plt.ylabel('امتیاز')
plt.ylim(0, 1)
plt.legend(loc='lower right')
plt.show()
```

## ۷. انتخاب مدل نهایی
مدل برتر را بر اساس **F1-Score** انتخاب می‌کنیم، چون F1 بین Precision و Recall تعادل برقرار می‌کند و برای مسئله پزشکی نامتوازن‌تر از Accuracy تنها قابل‌اعتمادتر است.
سپس Confusion Matrix و گزارش کامل طبقه‌بندی مدل برگزیده را بررسی می‌کنیم.

```python
best_model_name = results_df.iloc[0]['Model']
model_map = {
    'Logistic Regression': (log_reg, y_pred_lr),
    'KNN': (best_knn, y_pred_knn),
    'Decision Tree': (best_dt, y_pred_dt),
    'SVM': (best_svm, y_pred_svm),
}
final_model, final_preds = model_map[best_model_name]
print(f"مدل نهایی انتخاب‌شده: {best_model_name}")

print("\nClassification Report:")
print(classification_report(y_test, final_preds, target_names=['شانس کم', 'شانس زیاد']))

cm = confusion_matrix(y_test, final_preds)
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
            xticklabels=['پیش‌بینی: کم', 'پیش‌بینی: زیاد'],
            yticklabels=['واقعی: کم', 'واقعی: زیاد'])
plt.title(f'Confusion Matrix — {best_model_name}')
plt.show()
```

## ۸. ذخیره مدل نهایی
مدل برگزیده و `scaler` (که برای پیش‌پردازش داده‌های جدید لازم است) را با `joblib` ذخیره می‌کنیم تا بعداً بدون نیاز به آموزش دوباره قابل استفاده باشند.

```python
joblib.dump(final_model, 'best_model.pkl')
joblib.dump(scaler, 'scaler.pkl')
results_df.to_csv('model_comparison_results.csv', index=False)
print("مدل، اسکیلر و جدول مقایسه ذخیره شدند.")
```
